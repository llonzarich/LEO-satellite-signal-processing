```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
pd.options.plotting.backend = "plotly"
```

```python
def get_metrics(data):
    # total = [sum(d) for d in data]
    data2 = []
    for i in range(len(data)):
        new_data = []
        for j in range(len(data[i])):
            val = data[i][j]
            new_val = val/60
            new_data.append(new_val)
        data2.append(new_data)
    print(data2)
    # data2 = [d/6 for d in data]
    means = [np.mean(d) for d in data2]
    stds = [np.std(d) for d in data2]
    return (means, stds)
```

```python
# channel vs detection probability
xaxis = [1, 2, 3,  4, 5,  6,  7,  8]
ch8 = [8, 20, 14, 10, 16]
ch7 = [34, 33, 34, 33, 31]
ch6 = [20, 5]
ch5 = [54, 35, 42, 21, 22, 24]
ch4 = [3, 4, 2, 6, 4, 2]
ch3 = [3, 2, 3, 0, 8]
ch2 = [14, 11, 10, 11, 6]
ch1 = [32, 19, 17, 13, 17]
means, stds = get_metrics([ch1, ch2, ch3, ch4, ch5, ch6, ch7, ch8])
plt.errorbar(xaxis, means, yerr=stds, fmt='o', capsize=5, ecolor='red', elinewidth=2, markerfacecolor='blue', markersize=8)
plt.xlabel("channel")
plt.ylabel("detection probability")
# ax.set_title("My Simple Plot")
plt.ylim(ymin=0) 
plt.savefig("channel.pdf", format="pdf", bbox_inches="tight")
```

```python
# hit rate plot
# xticks fix: https://stackoverflow.com/questions/6682784/reducing-number-of-plot-ticks

df = pd.read_csv('/home/wairimu/Downloads/jul16.csv')
df = df.set_axis(['column_string'], axis=1)
DF_LENGTH = 10000

print(df.shape)
print(df.dtypes)
df = df[:DF_LENGTH]

df[['date', 'time', 'throughput']] = df['column_string'].str.split(' ', expand=True)
df = df.drop(['date', 'column_string'], axis=1)
print(df.dtypes)
df['throughput'] = pd.to_numeric(df['throughput'], downcast='integer', errors='coerce')
df['time'] = pd.to_datetime(df['time'],  format='%H:%M:%S')
# df['time'] = df['time'].dt.strftime('%H:%M:%S')
df['time'] = df['time'].dt.strftime('%H:%M')



df = df.groupby(['time']).mean()
# print(df.head(20))

constant_line = [200]*df.size

plt.scatter(df.index, df.throughput, label = 'actual throughput')
plt.scatter(df.index, constant_line, c='r', label = 'requested throughput')
plt.legend(loc='upper left')
plt.xlabel("time (hours)")
plt.ylabel("throughput (mbps)")
N = 30  # 1 tick every 3
xticks_pos, xticks_labels = plt.xticks()  # get all axis ticks
myticks = [j for i,j in enumerate(xticks_pos) if not i%N]  # index of selected ticks
newlabels = [label for i,label in enumerate(xticks_labels) if not i%N]
plt.xticks(myticks, newlabels)
plt.xticks(fontsize=10)
```
